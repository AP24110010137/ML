# **Matplotlib in Python**

Matplotlib is a powerful Python plotting library used for creating static, animated, and interactive visualizations.

It provides different types of plots such as:
- Line Plot
- Bar Plot
- Histogram
- Scatter Plot
- Pie Plot
- 3D Plot
- Contour Plot
- Polar Plot

## **Introduction to Matplotlib**

Matplotlib is popular because of its:
- Ease of use
- Extensive documentation
- Wide range of plotting capabilities

Other visualization libraries include:
- Seaborn
- Altair
- Bokeh
- Plotly

## **Importing Matplotlib**

The Matplotlib package can be imported using pyplot.

Pyplot is the most important module in Matplotlib for creating 2D plots.

```python
# AP24110010137

from matplotlib import pyplot as plt
```

## **Pyplot in Matplotlib**

Pyplot provides a MATLAB-like interface for creating plots.

It can be used to:
- Create a figure
- Create a plotting area
- Plot lines
- Add labels
- Add titles
- Customize the plot

## **Line Plot**

A line plot is created using the `plot()` function.

The `show()` function is used to display the graph.

```python
# AP24110010137

from matplotlib import pyplot as plt

x = [5, 2, 9, 4, 7]
y = [10, 5, 8, 4, 2]

plt.plot(x, y)

plt.savefig("line_plot.png")

plt.show()
```

```python
# AP24110010137

import matplotlib.pyplot as plt

X = [12, 34, 23, 45, 67, 89]
Y = [1, 3, 67, 78, 7, 5]

plt.plot(X, Y, marker='o', markerfacecolor='r')

plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")

plt.grid(linestyle='--', linewidth=0.5)

plt.title("Number of Users of a Particular Language")

plt.show()
```

```python
# AP24110010137

import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4, 5]
y = [3, 3, 3, 3, 3]

plt.plot(x, y, label="line 1", linestyle="-")
plt.plot(y, x, label="line 2", linestyle="--")
plt.plot(x, np.sin(x), label="curve 1", linestyle="-.")
plt.plot(x, np.cos(x), label="curve 2", linestyle=":")

plt.legend()

plt.show()
```

```python
# AP24110010137

import matplotlib.pyplot as plt
import pandas as pd

data = [
    ['A', 10, 20, 10, 26],
    ['B', 20, 25, 15, 21],
    ['C', 12, 15, 19, 6],
    ['D', 10, 18, 11, 19]
]

df = pd.DataFrame(
    data,
    columns=['Team', 'Round 1', 'Round 2', 'Round 3', 'Round 4']
)

print(df)

df.plot(
    x='Team',
    kind='bar',
    stacked=True,
    title='Stacked Bar Graph by DataFrame'
)

plt.show()
```

```python
# AP24110010137

import matplotlib.pyplot as plt

x = [1, 1.5, 2, 2.5, 3, 3.5, 3.6]
y = [7.5, 8, 8.5, 9, 9.5, 10, 10.5]

x1 = [8, 8.5, 9, 9.5, 10, 10.5, 11]
y1 = [3, 3.5, 3.7, 4, 4.5, 5, 5.2]

plt.scatter(x, y, label='High Income Low Saving')
plt.scatter(x1, y1, label='Low Income High Savings')

plt.xlabel('Saving * 100')
plt.ylabel('Income * 1000')
plt.title('Scatter Plot')

plt.legend()

plt.show()
```
