```python
import pandas as pd
import requests

# URL for a publicly available iris.csv dataset (e.g., from UCI ML Repo via a raw GitHub link)
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

try:
    # Download the file
    response = requests.get(url)
    response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)
    with open("iris.csv", "wb") as f:
        f.write(response.content)
    print("iris.csv downloaded successfully.")

    # Read the downloaded CSV file
    df1 = pd.read_csv('iris.csv')
    print("DataFrame head:")
    print(df1.head())
    print("\nDataFrame description:")
    print(df1.describe())
    print("\nDataFrame tail:")
    print(df1.tail())

except requests.exceptions.RequestException as e:
    print(f"Error downloading the file: {e}")
except FileNotFoundError:
    print("Error: 'iris.csv' not found after download attempt.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
